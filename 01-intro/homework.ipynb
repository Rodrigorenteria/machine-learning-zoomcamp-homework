```python
mediana_antes = df["horsepower"].median()
moda = df["horsepower"].mode()[0]

df["horsepower"] = df["horsepower"].fillna(moda)
mediana_despues = df["horsepower"].median()

print(mediana_antes, mediana_despues)
```

```python
X = (
    df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]]
    .head(7)
    .to_numpy()
)

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y
print(w.sum())
```
